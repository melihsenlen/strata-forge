```python
import numpy as np
import matplotlib.pyplot as plt

from engine.noise import select
from engine.mesh import normalize
from engine.config import load_config
```

```python
config = load_config("config.yaml")

terrain = config["terrain"]
noise = config["noise"]
```

```python
hm, seed = select(noise["type"], config)
hm = normalize(hm)
```

```python
fig = plt.figure(figsize=(6, 6))
ax = fig.add_subplot(111, projection="3d")

X, Y = np.meshgrid(
    range(terrain["width"]),
    range(terrain["height"])
)

ax.plot_surface(X, Y, hm, cmap="terrain")
ax.set_title(f"noise: {noise['type'].upper()} | seed: {seed}")
plt.show()
```
